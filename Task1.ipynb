Plasma Temperature Classifier

```python
# defining a function to classify the plasma based on the temperature 
```

```python
def classify_plasma_temp(temp):
    if temp < 1e4:                                # temp = temperature , 1e4 implies 1 x 10^4
       return "Cold plasma"
    elif 1e4 <= temp <= 1e6:
       return "Warm plasma"
    else:
       return "Hot plasma"
```

```python
data = []                                                            # An Empty list to store our input data
```

```python
# Asking the user to give the data 
```

```python
# Then simultaneously adding the user given data to the list data
```

```python
for i in range(5):
    try:
      t = float(input(f"Enter temperature {i+1} (K): "))            # Taking decimal input from the user 
      category = classify_plasma_temp(t)
      data.append((t, category))                                    # placing the data from user to our list named "data"
    except ValueError:
      print("Invalid input, please enter a number")                 # To convey the user that the input takes only numbers i.e.,throws error if i -
                                                                    # -give alphabets or special characters
```

```python
# displaying the classified data 
```

```python
# Print formatted table
print("\nPlasma Temperature Classification Table")
print("-" * 45)
print("{:<20} {:<20}".format("Temperature (K)", "Category"))
print("-" * 45)

for temp, category in data:
    print("{:<20.2e} {:<20}".format(temp, category))

print("-" * 45)
```

Temperature classification in plasma physics experiments is crucial because it dictates the plasma's physical properties, determines the appropriate diagnostic tools, and defines the feasibility of specific applications, ranging from fusion energy to industrial processing. Plasma is broadly classified by temperature into high-temperature (thermal/hot) plasma and low-temperature (non-thermal/cold) plasma, based on the equilibrium between electrons and heavy species (ions/neutrals). 

High-Temperature (Hot) Plasma: In these plasmas, electron, ion, and gas temperatures are nearly identical due to frequent collisions (local thermodynamic equilibrium or LTE). These are typically produced in fusion devices and are characterized by high conductivity, strong collective behavior, and intense electromagnetic interaction.

Low-Temperature (Cold) Plasma: Electrons are very hot (thousands of Kelvin), but ions and neutrals remain relatively cold (often near room temperature). This nonequilibrium state is maintained in electrical discharges, where energy is preferentially transferred to electrons.
