### Simple Particle Energy Calculator

Energy of relativistic particle: 

$ E =( \gamma - 1)mc^2$

$\gamma =\sqrt{1-\frac{v^2}{c^2}} $

```python
# To unlock a drawer full of mathematical tools 
```

```python
import math
```

```python
# Defining a class Particle with attributes mass,velocity 
class Particle:
    c = 3e8                                                                               # speed of light=3 x 10^8 (m/s)
    def __init__(q, name, m, v, charge=0.0):                                              # m = mass in kg, v = velocity in m/s 
        q.name = name
        q.m = float(m)
        q.v = float(v)
        q.charge = float(charge)
    def gamma(q):
        v = q.v
        c = Particle.c
        if v >= c:
            raise ValueError("Velocity cannot equal or exceed speed of light.")
        return 1 / math.sqrt(1 - (v**2 / c**2))
    def relativistic_kinetic_energy(q):
        gamma = q.gamma()
        return (gamma - 1) * q.m * Particle.c**2
    def classical_kinetic_energy(q):
        return 0.5 * q.m * q.v**2                                                         # Classical Energy E =(1/2) mv^2
    def potential_energy(q, V):
        return q.charge * float(V)
```

```python
# Conversion factor
```

```python
JeV = 1 / 1.6e-19
```

```python
# Creating 3 particle objects or List (e.g. electron, proton, alpha particle )
```

```python
# structure of particle function : Particle(name,mass,velocity,charge)
```

```python
particles = [
             Particle("Electron", 9.11e-31, 2e6, -1.6e-19),
             Particle("Proton", 1.67e-27, 1e5, 1.6e-19),
             Particle("Alpha", 6.64e-27, 5e4, 3.2e-19)
            ]
```

```python
# Loop and print kinetic energy in eV
```

```python
print("Particle Energies: \n")
for p in particles:
    KE_rel = p.relativistic_kinetic_energy()
    KE_class = p.classical_kinetic_energy()
    
    print(f"{p.name}")
    print(f"  Gamma (γ): {p.gamma():.4f}")
    print(f"  Classical KE: {KE_class * JeV:.5e} eV")
    print(f"  Relativistic KE: {KE_rel * JeV:.5e} eV")
```

```python
V = 1000                                                                                  # 1000 Volts
print("Electron PE at 1000 V:", f"{particles[1].potential_energy(V)*JeV:.2e}", "eV")
```
