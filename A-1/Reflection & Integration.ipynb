## Python in Astrophysics

Astrophysics swims in oceans of data. Telescopes like NASA’s James Webb Space Telescope and surveys from European Space Agency stream terabytes of images and spectra daily. Python acts as the observatory’s quiet engine room. With libraries for numerical analysis, visualization, and machine learning, researchers classify galaxies, detect exoplanets, and model stellar evolution efficiently. Data cleaning, statistical inference, and simulation all live inside one flexible ecosystem, which means a student analyzing star temperatures and a scientist modeling relativistic jets often use the same language.

### Quick Reflection

The easiest part was structuring physics formulas into functions and classes. Translating $ KE = \frac{1}{2}mv^2 $ or $ E =( \gamma - 1)mc^2$ into Python feels natural because of the syntax closely mirrors mathematical notation. Once the formula is clear  implementation becomes easy.

The hardest part was managing formatting and unit consistency, especially when converting between Joules and eV or presenting values in clean scientific notation. Tiny formatting details, like controlling $1.00e+03$ vs $1.00000000e+03$ something like that, require careful attention, and mistakes in units can quietly distort results. Anaconda environments are extremely helpful in real research because they isolate project dependencies. In astrophysics different projects require specific versions of NumPy, SciPy, or specialized libraries. Environments prevent version conflicts and make experiments reproducible
