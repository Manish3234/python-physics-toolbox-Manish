### Material Properties Lookup

```python
# Creating a dictionary 'materials' having all the info about density, conductivity and youngs modulus of certain materials
```

```python
materials = {
             "Aluminum": {"density_g_cm3": 2.7, "conductivity": 3.5e7, "youngs_modulus_GPa": 70},
             "Copper": {"density_g_cm3": 8.96, "conductivity": 5.9e7, "youngs_modulus_GPa": 110},
             "Silicon": {"density_g_cm3": 2.33, "conductivity": 1.5e-3, "youngs_modulus_GPa": 130},
             "Graphene": {"density_g_cm3": 2.26, "conductivity": 1e8, "youngs_modulus_GPa": 1000}
            }
```

```python
# defining a function that carries out adding new material and its data 
```

```python
def add_material(name, density, conductivity, youngs):
    materials[name] = {
                       "density_g_cm3": density,
                       "conductivity": conductivity,
                       "youngs_modulus_GPa": youngs
                      }
```

```python
# a function for finding the material with highest conductivity.
```

```python
def highest_conductivity():
    tp = []
    for i in materials:
        tp.append(materials[i]["conductivity"])

    x = None   
    
    for p in materials:
        if materials[p]["conductivity"] == max(tp):
            x = p
            
    return x
```

```python
# function to •	Return list of densities as floats.
```

```python
def density_list():
    tp =[]
    for i in materials:
        tp.append(materials[i]["density_g_cm3"])
    return tp
```

```python
# a function for finding the material with highest density.
```

```python
def highest_density():
    tp = []
    for i in materials:
        tp.append(materials[i]["density_g_cm3"])

    x = None   
    
    for p in materials:
        if materials[p]["density_g_cm3"] == max(tp):
            x = p
            
    return x
```

```python
# displaying material havong highest conductivity and list of densities of the materials
```

```python
print("Material with highest conductivity:", highest_conductivity())
print("List of densities:", density_list())
```

```python
# if i want to add "Gold" to this dicitonary having density= 19.3 g/cm^3 conductivity = 4.5 x 10^7 S/m youngs modulus = 78 GPa
```

```python
add_material("Gold",19.3,4.5e7,78)
print("Updated list of density :",density_list())
```

```python
# Let's ask the material that is having highest density now
```

```python
print("Material with highest conductivity:",highest_density())
```

### Why Dictionaries Are Powerful for Experimental Data ?

Dictionaries (or similar key value mapping structures) are used in physics to organize experimental data by providing a flexible, human readable, and machine interpretable way to pair metadata (labels, units) with numerical values. They enable efficient data handling, particularly for managing complex, multi dimensional, or sparse data sets commonly found in experimental physics.
